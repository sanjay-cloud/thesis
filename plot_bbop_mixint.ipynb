```python
import cocopp
import cocoex
import cma
import numpy as np
import matplotlib.pyplot as plt
import scipy.optimize
from numpy.random import rand  
```

```python
def get_problem(suite_name = "bbob-mixint",suite_num = 0):
    suite = cocoex.Suite(suite_name, "", "")
    return suite[suite_num]
```

```python
def opt_problem(problem, fmin = scipy.optimize.fmin):
    x0 = problem.initial_solution
    while (not problem.final_target_hit):
        f_opt = fmin(problem, x0, disp=False)
        # xopt, es = fmin(problem, x0, 2,restarts=9)
        x0 = problem.lower_bounds + ((rand(problem.dimension) + rand(problem.dimension)) *
                    (problem.upper_bounds - problem.lower_bounds) / 2)
    return f_opt    
```

```python
def bbop_mixint_prob(problem, fopt, x, y):
    # fopt = np.array([0, 0, 0, 0, 0])
    fopt[3] = x
    fopt[4] = y
    return problem(fopt)
```

```python
def get_results(problem, fopt):
    lower_bounds = problem.lower_bounds[3:]
    upper_bounds = problem.upper_bounds[3:]

    continuous_var_range = np.linspace(lower_bounds[1], upper_bounds[1], 100)  
    discrete_var_range = np.arange(lower_bounds[0], upper_bounds[0], 1) 
    results = np.zeros((len(discrete_var_range), len(continuous_var_range)))
    
    for i, discrete_val in enumerate(discrete_var_range):
        for j, continuous_val in enumerate(continuous_var_range):
            results[i, j] = bbop_mixint_prob(problem, fopt, discrete_val, continuous_val)

    return  np.meshgrid(continuous_var_range, discrete_var_range) , results
```

```python
def plot_results():
    suite_num = 0
    while(suite_num < 1):
        problem = get_problem(suite_num = suite_num)
        fopt = opt_problem(problem)

        (X, Y), results = get_results(problem, fopt)
        # print(np.argmin(results))
        # print(results[81][8])
        # print(problem.best_observed_fvalue1)
        plt.figure(figsize=(5, 4))
        cp = plt.contour(Y, X, results, levels=20, cmap='viridis')
        plt.clabel(cp, inline=True, fontsize=10, fmt='%1.1f')  # Add labels to the contour lines
        plt.colorbar(cp)
        plt.xlabel('Discrete  variable (x)')
        plt.ylabel('Continuous variable (y)')
        plt.title('Level Sets of Function Values')
        plt.xticks([0, 5, 10, 15])  # Set x-axis ticks to 0, 5, 10, 15
        name = problem.name
        plt.savefig("pics/"+name+".png")
        suite_num += 15
```

```python
plot_results()
```
